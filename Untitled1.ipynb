```python
import tkinter as tk
from tkinter import scrolledtext
import datetime
import webbrowser
import pyttsx3

# Voice Engine
engine = pyttsx3.init()

def speak(text):
    engine.say(text)
    engine.runAndWait()

# Assistant Logic
def assistant_reply(user_input):
    user_input = user_input.lower()

    if "hello" in user_input:
        return "Hello Bro 😎"

    elif "time" in user_input:
        current_time = datetime.datetime.now().strftime("%I:%M %p")
        return f"Current time is {current_time}"

    elif "youtube" in user_input:
        webbrowser.open("https://youtube.com")
        return "Opening YouTube"

    elif "google" in user_input:
        webbrowser.open("https://google.com")
        return "Opening Google"

    elif "github" in user_input:
        webbrowser.open("https://github.com")
        return "Opening GitHub"

    elif "bye" in user_input:
        return "Goodbye Bro 👋"

    else:
        return "Sorry Bro, samaj nahi aya 😅"

# Send Message
def send_message():
    user_input = entry.get()

    if user_input.strip() == "":
        return

    chat_area.insert(tk.END, f"You: {user_input}\n")

    reply = assistant_reply(user_input)

    chat_area.insert(tk.END, f"Assistant: {reply}\n\n")

    speak(reply)

    entry.delete(0, tk.END)

# GUI Window
window = tk.Tk()
window.title("Smart AI Assistant")
window.geometry("500x600")
window.config(bg="#1e1e1e")

# Title
title = tk.Label(
    window,
    text="🤖 Smart AI Assistant",
    font=("Arial", 20, "bold"),
    bg="#1e1e1e",
    fg="white"
)
title.pack(pady=10)

# Chat Area
chat_area = scrolledtext.ScrolledText(
    window,
    wrap=tk.WORD,
    width=55,
    height=25,
    font=("Arial", 11),
    bg="#2d2d2d",
    fg="white"
)
chat_area.pack(padx=10, pady=10)

# Input Frame
frame = tk.Frame(window, bg="#1e1e1e")
frame.pack(pady=10)

entry = tk.Entry(
    frame,
    width=35,
    font=("Arial", 14)
)
entry.pack(side=tk.LEFT, padx=5)

send_btn = tk.Button(
    frame,
    text="Send",
    font=("Arial", 12, "bold"),
    bg="#4CAF50",
    fg="white",
    command=send_message
)
send_btn.pack(side=tk.LEFT)

window.mainloop()
```
